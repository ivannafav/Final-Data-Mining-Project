# **Models**

```python
#Splitting for train and test

train_test_data = {}

for col in selected_Ys:

    TS = df[col].dropna().reset_index(drop=True)

    train_size = int(len(TS) * 0.8)
    train = TS.iloc[:train_size]
    test = TS.iloc[train_size:]

    train_test_data[col] = (train, test)

    print(f"\n{col}")
    print('Train size:', len(train))
    print('Test size:', len(test))

    plt.figure(figsize=(8,4))
    plt.plot(train, label='Train')
    plt.plot(test, label='Test')
    plt.title(f'{col} Split')
    plt.legend()
    plt.show()
```

## ARIMA

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error
from statsmodels.tsa.arima.model import ARIMA
```

### Y1

```python
train, test = train_test_data['Y1']

results_Y1 = []

#Model

model = ARIMA(train, order=(2,1,2))
model_fit = model.fit()

print(model_fit.summary())

#Plot fitted vs. actual

fitted_values = model_fit.predict(start=train.index[0], end=train.index[-1], dynamic=False)

plt.figure(figsize=(10,6))
plt.plot(train, label='Actual')
plt.plot(fitted_values, label='Fitted')
plt.title('Actual vs Fitted (Training Set) - Y1')
plt.legend()
plt.show()

#Forecast test set

arima_forecast = model_fit.forecast(steps=len(test))

print("len(arima_forecast):", len(arima_forecast))

#Converts metrics to arrays

y_true = test.values
y_pred = arima_forecast.values

#Metrics

rmse = np.sqrt(mean_squared_error(y_true, y_pred))
mae = mean_absolute_error(y_true, y_pred)
mad = mae
mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100

print("RMSE:", rmse)
print("MAE:", mae)
print("MAD:", mad)
print("MAPE:", mape)

#Save for later

results_Y1.append({
    'Model': 'ARIMA',
    'RMSE': rmse,
    'MAE': mae,
    'MAD': mad,
    'MAPE': mape
})
```

```python
#Results for Excel

arima_results_Y1 = pd.DataFrame({
    'True': test.values,
    'ARIMA': arima_forecast.values
}, index=test.index)
```

## Random Forest

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
#Lagged dataset (30)

def create_lagged_data(series, n_lags=30):
    lagged_df = pd.DataFrame()

    for i in range(1, n_lags + 1):
        lagged_df[f'lag_{i}'] = series.shift(i)

    lagged_df['target'] = series

    return lagged_df.dropna()
```

### Y1

```python
#Applying data to Y1

train, test = train_test_data['Y1']

full_Y1 = pd.concat([train, test], ignore_index=True)
lagged_Y1 = create_lagged_data(full_Y1, n_lags=30)

#Train/test split

train_lag = lagged_Y1.iloc[:len(train) - 30]
test_lag = lagged_Y1.iloc[len(train) - 30:]

#X and y

X_train = train_lag.drop(columns=['target'])
y_train = train_lag['target']

X_test = test_lag.drop(columns=['target'])
y_test = test_lag['target']

#RF

rf_model = RandomForestRegressor(
    n_estimators=300,
    max_depth=10,
    min_samples_leaf=5,
    random_state=42
)

rf_model.fit(X_train, y_train)

#Predict

rf_pred = rf_model.predict(X_test)

#Plot fitted vs. actual

plt.figure(figsize=(10,5))
plt.plot(y_test.index, y_test, label='Actual')
plt.plot(y_test.index, rf_pred, label='Predicted')
plt.title('Y1: Actual vs Predicted (Random Forest)')
plt.legend()
plt.show()

#Metrics

rmse_rf = np.sqrt(mean_squared_error(y_test, rf_pred))
mae_rf = mean_absolute_error(y_test, rf_pred)
mad_rf = mae_rf
mape_rf = np.mean(np.abs((y_test - rf_pred) / y_test)) * 100


print("RMSE:", rmse_rf)
print("MAE:", mae_rf)
print("MAD:", mad_rf)
print("MAPE:", mape_rf)

#Results for later

results_Y1.append({
    'Model': 'RF',
    'RMSE': rmse_rf,
    'MAE': mae_rf,
    'MAD': mad_rf,
    'MAPE': mape_rf
})

#Results for excel

rf_results_Y1 = pd.DataFrame({
    'True': y_test.values,
    'RF': rf_pred
}, index=y_test.index)
```
